```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

urls = [
    "https://sece.ac.in/admissions/",
    "https://sece.ac.in/academics/",
    "https://sece.ac.in/departments/",
    "https://sece.ac.in/recruiters/",
    "https://sece.ac.in/global-alliances/",
    "https://sece.ac.in/governance/",

    "https://sece.ac.in/department-mech/",
    "https://sece.ac.in/department-aids/",
    "https://sece.ac.in/department-computer-communication-engineering/"
]

data = []

for url in urls:
    try:
        response = requests.get(url)

        soup = BeautifulSoup(response.text, "html.parser")

        title = soup.title.text if soup.title else "No Title"

        text = soup.get_text(separator=" ", strip=True)

        clean_text = " ".join(text.split())

        data.append({
            "url": url,
            "title": title,
            "text": clean_text
        })

        print(f"Done: {url}")

    except Exception as e:
        print(f"Error: {url}")
        print(e)

df = pd.DataFrame(data)

df.head()
```

```python
df.to_csv("sece_knowledge_base.csv", index=False)

print("CSV saved successfully!")
```

```python
import pandas as pd

df = pd.read_csv("sece_knowledge_base.csv")

chunks = []

chunk_size = 500

for _, row in df.iterrows():

    text = row["text"]

    for i in range(0, len(text), chunk_size):

        chunk = text[i:i+chunk_size]

        chunks.append({
            "source_url": row["url"],
            "title": row["title"],
            "chunk_text": chunk
        })

chunk_df = pd.DataFrame(chunks)

print("Total Chunks:", len(chunk_df))

chunk_df.head()
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

data = []

for url in urls:

    try:

        response = requests.get(url)

        soup = BeautifulSoup(response.text, "html.parser")

        title = soup.title.text if soup.title else "No Title"

        text = soup.get_text(separator=" ", strip=True)

        clean_text = " ".join(text.split())

        data.append({
            "url": url,
            "title": title,
            "text": clean_text
        })

        print("Done:", url)

    except Exception as e:

        print("Error:", url)
        print(e)

df = pd.DataFrame(data)

print(df.shape)
df.head()
```

```python
def chunk_text(text, chunk_size=100):

    words = text.split()

    chunks = []

    for i in range(0, len(words), chunk_size):

        chunk = " ".join(words[i:i + chunk_size])

        chunks.append(chunk)

    return chunks
```

```python
chunks = []

for _, row in df.iterrows():

    page_chunks = chunk_text(
        row["text"],
        chunk_size=100
    )

    for chunk in page_chunks:

        chunks.append({
            "source_url": row["url"],
            "title": row["title"],
            "chunk_text": chunk
        })

import pandas as pd

chunk_df = pd.DataFrame(chunks)

print(chunk_df.shape)

chunk_df.head()
```

```python
texts = chunk_df["chunk_text"].tolist()

embeddings = model.encode(
    texts,
    show_progress_bar=True
)

print(embeddings.shape)
```

```python
import faiss
import numpy as np

embedding_array = np.array(
    embeddings,
    dtype="float32"
)

dimension = embedding_array.shape[1]

index = faiss.IndexFlatL2(dimension)

index.add(embedding_array)

print("Vectors stored:", index.ntotal)
```

```python
def search_chunks(query, k=5):

    query_embedding = model.encode([query])

    D, I = index.search(
        np.array(query_embedding, dtype="float32"),
        k=k
    )

    results = []

    for idx in I[0]:
        results.append({
            "text": chunk_df.iloc[idx]["chunk_text"],
            "url": chunk_df.iloc[idx]["source_url"]
        })

    return results
```

```python
results = search_chunks(
    "What are the admission requirements?"
)

for r in results:
    print("\n" + "="*80)
    print(r)
```

```python
def ask_sece(question):

    retrieved = search_chunks(question, k=5)

    context = "\n\n".join(
        [item["text"] for item in retrieved]
    )

    sources = list(set(
        [item["url"] for item in retrieved]
    ))

    prompt = f"""
You are the official SECE Information Assistant.

Answer ONLY from the provided context.

If the answer is not present in the context, reply exactly:

'I could not find that information in the SECE knowledge base.'

Context:
{context}

Question:
{question}

Answer:
"""

    response = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ],
        temperature=0.2
    )

    answer = response.choices[0].message.content

    answer += "\n\nSources:\n"

    for source in sources:
        answer += f"- {source}\n"

    return answer
```
